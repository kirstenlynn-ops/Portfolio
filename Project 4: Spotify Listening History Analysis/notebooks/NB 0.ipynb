# Initial Analysis of Spotify Listening History Data
---
**Goal:**
- Load & inspect datasets for both users (2021-2025)
- Convert raw JSON data to a Pandas DataFrame
- Understand basic structure, column types, & data completeness
- Record any issues or early insights for downstream analysis

**Tasks:**

Step 1: User A
1. Import libraries
2. Load raw Spotify JSON files (2021-2025)
3. Inspect data:

   - Time coverage/ Overlap
   - Column names & types
   - Missing values

Step 2: User B
1. Repeat steps 2-7 in Step 1

Step 3: Final | Readiness Check

2. Ensure data is ready for NB 1

```python
#Libraries
import os
import json
import pandas as pd
```

## User A | Load
---

```python
#User A raw data folder
user_a_folder = "User A Raw Data/"

#List all JSON files in folder (sort by filename)
user_a_files = sorted([f for f in os.listdir(user_a_folder) if f.endswith(".json")])

#Individual df's
user_a_dfs = []

#Loop each file, load, & inspect
for file in user_a_files:
    file_path = os.path.join(user_a_folder, file)
    
    with open(file_path, "r", encoding="utf-8") as f:
        data = json.load(f)
    
    df = pd.DataFrame(data)
    
#Basic inspection
    print(f"--- File: {file} ---")
    print("Shape:", df.shape)
    print("Columns:", df.columns.tolist())
    print("Time range:", df["ts"].min(), "to", df["ts"].max())
    print("Missing values per column:")
    print(df.isnull().sum())
    print("\n")
    
    user_a_dfs.append(df)

#Combine all files into one df
df_user_a = pd.concat(user_a_dfs, ignore_index=True)

#Convert timestamp to dt
df_user_a['ts'] = pd.to_datetime(df_user_a['ts'], errors='coerce')

#Reset index
df_user_a.reset_index(drop=True, inplace=True)

#Save combined raw data
df_user_a.to_csv("User A Raw Combined/User_A_Raw_2021-2025.csv", index=False)

print(f"User A | Combined shape: {df_user_a.shape}")
print(f"Time range: {df_user_a['ts'].min()} to {df_user_a['ts'].max()}")
```

## User B | Load
---

```python
#Repeat for User B

#User B raw data folder
user_b_folder = "User B Raw Data/"

#List all JSON files in folder (sort by filename)
user_b_files = sorted([f for f in os.listdir(user_b_folder) if f.endswith(".json")])

#Hold the individual df's
user_b_dfs = []

#Loop through each file, load, & inspect
for file in user_b_files:
    file_path = os.path.join(user_b_folder, file)
    
    with open(file_path, "r", encoding="utf-8") as f:
        data = json.load(f)
    
    df = pd.DataFrame(data)
    
#Basic inspection for User B
    print(f"--- File: {file} ---")
    print("Shape:", df.shape)
    print("Columns:", df.columns.tolist())
    print("Time range:", df["ts"].min(), "to", df["ts"].max())
    print("Missing values per column:")
    print(df.isnull().sum())
    print("\n")
    
    user_b_dfs.append(df)

#Combine all files into one df
df_user_b = pd.concat(user_b_dfs, ignore_index=True)

#Convert timestamp to dt
df_user_b['ts'] = pd.to_datetime(df_user_b['ts'], errors='coerce')

#Reset index
df_user_b.reset_index(drop=True, inplace=True)

#Save combined raw data
df_user_b.to_csv("User B Raw Combined/User_B_Raw_2021-2025.csv", index=False)

print(f"User B | Combined shape: {df_user_b.shape}")
print(f"Time range: {df_user_b['ts'].min()} to {df_user_b['ts'].max()}")
```

## Observations
---

Shape:

- Consistent 23 columns

Time Coverage:
- Start: 11/03/2021
- End: 09/07/2025
- Covers 2021-2025

Missing Data:

Notes:

**User B:**

Shape:
- Files range ~15,900-16,100 rows
- Consistent 23 columns

Time Coverage:
- Start: 05/12/2021
- End: 08/04/2025
- Covers a little more range than User A (at the start)

Notes:
- `offline_timestamp` is mostly missing in older folders

## Schema & Compatibility Check
---

```python
#Compare:

#Column names
print("Columns in User A:")
print(df_user_a.columns.tolist())
print("\nColumns in User B:")
print(df_user_b.columns.tolist())

#Data types
print("\nData types in User A:")
print(df_user_a.dtypes)
print("\nData types in User B:")
print(df_user_b.dtypes)

#Missing values
print("\nMissing values in User A:")
print(df_user_a.isnull().sum())
print("\nMissing values in User B:")
print(df_user_b.isnull().sum())

#Highlight differences in columns
cols_a = set(df_user_a.columns)
cols_b = set(df_user_b.columns)
print("\nColumns only in User A:", cols_a - cols_b)
print("Columns only in User B:", cols_b - cols_a)
```
